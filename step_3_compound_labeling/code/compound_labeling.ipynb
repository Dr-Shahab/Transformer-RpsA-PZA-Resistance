```python
from rdkit import Chem
import pandas as pd
import numpy as np
```

```python
molecules  = pd.read_csv(r"E:\Projects\Mycobacterium tuberculosis\02-Generate Standard Smile\input\bio_12756.csv")
```

```python
molecules
```

```python
print(molecules.isnull().sum())
```

```python
def convert_to_valid_smiles(smiles):
    mol = Chem.MolFromSmiles(smiles)
    if mol is not None:
        return Chem.MolToSmiles(mol)
    else:
        return None
```

```python
molecules['Standard_smiles'] = molecules['Smiles'].apply(convert_to_valid_smiles)
```

```python
molecules.shape
```

```python
molecules.head()
```

```python
valid_molecules = molecules[molecules['Standard_smiles'].notnull()]
```

```python
valid_molecules
```

```python
valid_molecules = valid_molecules.drop_duplicates("Standard_smiles", keep="first")
```

```python
valid_molecules
```

```python
valid_molecules.shape
```

```python
valid_molecules = pd.DataFrame(
    {'Molecule ChEMBL ID':valid_molecules.loc[:,'Molecule ChEMBL ID'],
     'MIC':valid_molecules.loc[:,'MIC'],
     'Units':valid_molecules.loc[:,'Units'],
     'Smiles':valid_molecules.loc[:,'Standard_smiles']
    }
)
```

```python
valid_molecules
```

```python
min_value = valid_molecules['MIC'].min()
max_value = valid_molecules['MIC'].max()

median_value = valid_molecules["MIC"].median()
mean_value = valid_molecules["MIC"].mean()
```

```python
def classify_resistance(row):
    #drug = str(row["Molecule Name"]).upper()
    mic = row["MIC"]
    return 0 if mic >= median_value else 1
```

```python
valid_molecules["Label"] = valid_molecules.apply(classify_resistance, axis=1)
```
