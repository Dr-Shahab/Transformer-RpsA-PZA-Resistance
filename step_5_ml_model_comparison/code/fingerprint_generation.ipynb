```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from rdkit import Chem, DataStructs
from rdkit.Chem import (
    PandasTools,
    Draw,
    Descriptors,
    MACCSkeys,
    rdFingerprintGenerator,
    rdMolDescriptors
)

import matplotlib.pyplot as plt
```

```python
molecules  = pd.read_csv(r"E:\Projects\Mycobacterium tuberculosis\ML performance comparison\Input files\te_df_64.csv")
```

```python
molecules.head()
```

```python
PandasTools.AddMoleculeColumnToFrame(molecules, smilesCol="Smiles")
```

```python
# get ecfp4 FP
fpg = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=1024)
molecules["ecfp4"] = molecules["ROMol"].apply(lambda m: fpg.GetFingerprint(m)) #bit
```

```python
print(len(molecules["ecfp4"].iloc[0]))
```

```python
ecfp4_df = pd.DataFrame([list(i.ToBitString()) for i in molecules["ecfp4"]], columns=[f"ECFP4_{i}" for i in range(1024)])
molecules_ecfp4 = pd.concat([molecules.iloc[:,:5], ecfp4_df], axis=1)
```
