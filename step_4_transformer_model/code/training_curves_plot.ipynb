```python
import pandas as pd
import numpy as np

import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import confusion_matrix

plt.rcParams['font.family'] = 'serif' # 用 serif 字体（字体大类）
plt.rcParams['font.serif'] = ['Times New Roman'] # serif 具体用 Times New Roman
plt.rcParams['axes.unicode_minus'] = False
matplotlib.rcParams['pdf.fonttype'] = 42
matplotlib.rcParams['ps.fonttype'] = 42
```

```python
tr_df = pd.read_csv(r"E:\Projects\Mycobacterium tuberculosis\03-Machine Learning\result\8\Hyperparameters Rank 3\performance_tr_8.csv",usecols=["tr_loss","tr_mcc"])
val_df = pd.read_csv(r"E:\Projects\Mycobacterium tuberculosis\03-Machine Learning\result\8\Hyperparameters Rank 3\performance_val_8.csv",usecols=["val_loss","val_mcc"])
```

# loss curve

```python
train_loss = tr_df['tr_loss']
val_loss = val_df['val_loss']
epochs = range(len(train_loss))
```

```python
plt.figure(figsize=(8,6))

plt.plot(epochs, train_loss, marker='o', color='blue', linestyle='-', linewidth=2, label='Training Loss')

plt.plot(epochs, val_loss, marker='o', color='red', linestyle='-', linewidth=2, label='Validation Loss')


plt.xlabel('Epochs', fontsize=24, fontweight='bold')
plt.ylabel('Loss', fontsize=24, fontweight='bold')
plt.title('Loss Curves: Seed 8, Rank 3', fontsize=28, fontweight='bold')

plt.legend(
           frameon=False, 
           prop={'size':20,'weight':'bold'}
          )

max_epoch = len(train_loss)
xticks = list(range(0, max_epoch+1, 10))
plt.xticks(ticks=xticks, fontsize=20, fontweight='bold')
plt.yticks(fontsize=20, fontweight='bold')

ax = plt.gca()
for spine in ax.spines.values():
    spine.set_linewidth(1)       
    spine.set_color('black')     

plt.savefig("Loss Curves Seed 8 Rank 3.pdf", format='pdf', bbox_inches='tight')

plt.show()
```

# mcc curve

```python
train_mcc = tr_df['tr_mcc']
val_mcc = val_df['val_mcc']
epochs = range(len(train_mcc))
```

```python
plt.figure(figsize=(8,6))

plt.plot(epochs, train_mcc, marker='o', color='blue', linestyle='-', linewidth=2, label='Training MCC')

plt.plot(epochs, val_mcc, marker='o', color='red', linestyle='-', linewidth=2, label='Validation MCC')


plt.xlabel('Epochs', fontsize=24, fontweight='bold')
plt.ylabel('MCC', fontsize=24, fontweight='bold')
plt.title('MCC Curves: Seed 8, Rank 3', fontsize=28, fontweight='bold')

plt.legend(
           frameon=False, 
           prop={'size':20,'weight':'bold'}
          )

max_epoch = len(train_mcc)
xticks = list(range(0, max_epoch+1, 10))
plt.xticks(ticks=xticks, fontsize=20, fontweight='bold')
plt.yticks(fontsize=20, fontweight='bold')


ax = plt.gca()
for spine in ax.spines.values():
    spine.set_linewidth(1)
    spine.set_color('black')

plt.savefig("MCC Curves Seed 8 Rank 3.pdf", format='pdf', bbox_inches='tight')

plt.show()
```
